```python
import joblib
from sklearn.pipeline import Pipeline
```

```python
knn = joblib.load("models/tuned_knn.pkl")
```

```python
scaler = joblib.load("models/scaler.pkl")
```

```python
pipeline = Pipeline([
    ("scaler", scaler),
    ("knn", knn)
])
```

```python
joblib.dump(
    pipeline,
    "models/fraud_detection_pipeline.pkl"
)
```

```python
loaded_pipeline = joblib.load(
    "models/fraud_detection_pipeline.pkl"
)

print(loaded_pipeline)
```
